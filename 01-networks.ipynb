# Co-appearance of works in manuscripts — French vs German

Two works are linked whenever they share a manuscript; edge weight is the number of shared manuscripts. This yields a co-appearance network per tradition, used to test whether shared *matiere* (Britain, France, Rome) structures co-transmission, and whether French and German behave differently.

A work's date is its date of composition; a manuscript's date is that of the surviving copy. The two are never mixed.

```python
import json
from collections import Counter
from itertools import combinations
import numpy as np
import pandas as pd
import networkx as nx
from networkx.algorithms.community import greedy_modularity_communities
from sklearn.metrics import adjusted_rand_score, adjusted_mutual_info_score
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D

plt.rcParams["font.family"] = "sans-serif"
plt.rcParams["font.sans-serif"] = ["Roboto Condensed", "Roboto", "DejaVu Sans"]
plt.rcParams["figure.dpi"] = 150

INK, SUBTLE, GRID = "#111111", "#333333", "#e0e0e0"
LANGS = ["french", "german"]
MAT_ORDER_ALL = ["Britain", "France", "Rome", "Other", "Unknown"]
MAT_ORDER = ["Britain", "France", "Rome"]  # matieres retained for network/statistical analysis
MAT_COLORS = {"Britain": "#2E86AB", "France": "#E63946", "Rome": "#F4A261",
              "Other": "#2A9D8F", "Unknown": "#A0A0A0"}

def style_ax(ax, grid_axis="y"):
    ax.set_facecolor("white")
    for s in ("top", "right"):
        ax.spines[s].set_visible(False)
    for s in ("left", "bottom"):
        ax.spines[s].set_color(SUBTLE); ax.spines[s].set_linewidth(0.8)
    ax.tick_params(colors=SUBTLE, labelsize=8, length=3, width=0.5)
    if grid_axis:
        ax.grid(axis=grid_axis, color=GRID, alpha=0.6, lw=0.5); ax.set_axisbelow(True)
    return ax
```

## Building the network

`load_linkage` reads and cleans a linkage file; `build_graph` adds one node per work and one edge per pair of works sharing a manuscript, weighted by the number of shared manuscripts, after dropping fragmentary witnesses. Nodes are also tagged with a data-driven community (greedy modularity), used later as a yardstick against the matiere grouping.

```python
def load_linkage(path):
    """Read a linkage file. Its values are already analysis-ready -- titles
    unquoted, status normalised, matiere resolved -- see docs/PREPROCESSING.md.
    Only the dtypes JSON cannot carry need restoring."""
    with open(path, encoding="utf-8") as f:
        df = pd.DataFrame(json.load(f))
    df["manuscript_id"] = pd.to_numeric(df["manuscript_id"], errors="coerce").astype("Int64")
    # midpoint of the copying range, derived rather than stored (as for works)
    df["date_mid"] = ((pd.to_numeric(df["date_start"], errors="coerce")
                       + pd.to_numeric(df["date_end"], errors="coerce")) / 2).round()
    return df

def build_graph(df, weight="weight", exclude_status=("Fragmentary",)):
    if "status" in df and exclude_status:
        df = df[~df["status"].isin(exclude_status)].copy()

    G = nx.Graph()
    for wid, g in df.groupby("work_id", dropna=True):
        G.add_node(g.iloc[0]["work"], work_id=int(wid))
    for _, g in df.dropna(subset=["manuscript_id"]).groupby("manuscript_id"):
        for a, b in combinations(sorted(g["work"].dropna().unique()), 2):
            if G.has_edge(a, b):
                G[a][b][weight] += 1
            else:
                G.add_edge(a, b, **{weight: 1})

    comms = (greedy_modularity_communities(G, weight=weight)
             if G.number_of_edges() else [{n} for n in G.nodes()])
    nx.set_node_attributes(G, {n: i for i, c in enumerate(comms) for n in c}, "community")
    return G, df
```

## Loading both traditions and assigning matiere

`matiere` is already final in `{lang}_works.xlsx` -- resolved from the storyverse hierarchy, aliased, and with any manual corrections folded in during preprocessing (see `docs/PREPROCESSING.md`, step 9). No alias/override logic is needed here any more: `assign_matiere` just reads the column. `load_all` builds both graphs and returns the graphs, the filtered linkage tables, the works metadata, and the raw (pre-status-filter) linkage tables -- the latter needed below for manuscript preservation states.


```python
def assign_matiere(G, w, id_col="work_id", mat_col="matiere"):
    """matiere is final in the works file: one category per work, already
    aliased and corrected (see docs/PREPROCESSING.md). Nothing is remapped
    here -- this notebook only reads."""
    mm = {int(r[id_col]): r[mat_col] for _, r in w.iterrows()}
    nx.set_node_attributes(G, {n: mm.get(G.nodes[n].get("work_id"), "Unknown") for n in G.nodes()}, "matiere")

def tag_heldenepik(G, w, id_col="work_id", flag_col="is_heldenepik"):
    """Tag every node with is_heldenepik (bool) and mat5 (= matiere, but with
    Heldendichtung carved out of Other for flagged works). Language-agnostic: if the
    works file has no is_heldenepik column (e.g. French), mat5 simply mirrors matiere
    and nothing is flagged. Doing this at load time makes the Heldendichtung labelling
    available to the matiere-distribution figure below, not only to the German-specific
    split analysis much further down."""
    # pd.notna guard: a missing flag must read as False, not as bool(nan) == True
    flag = ({int(r[id_col]): bool(r[flag_col]) and pd.notna(r[flag_col])
             for _, r in w.iterrows()} if flag_col in w.columns else {})
    for n in G.nodes():
        h = flag.get(G.nodes[n].get("work_id"), False)
        m = G.nodes[n]["matiere"]
        G.nodes[n]["is_heldenepik"] = h
        G.nodes[n]["mat5"] = "Heldendichtung" if (h and m == "Other") else m

def load_all(langs, data="data"):
    graphs, links, works, raw = {}, {}, {}, {}
    for l in langs:
        raw_df = load_linkage(f"{data}/{l}_linkage.json")
        G, df = build_graph(raw_df)
        w = pd.read_excel(f"{data}/{l}_works.xlsx")
        # midpoint of the composition range; stored as start/end only, since
        # date_mid is exactly their mean and would just be a third copy.
        w["date_mid"] = ((pd.to_numeric(w["date_start"], errors="coerce")
                          + pd.to_numeric(w["date_end"], errors="coerce")) / 2).round()
        assign_matiere(G, w)
        tag_heldenepik(G, w)
        graphs[l], links[l], works[l], raw[l] = G, df, w, raw_df
    return graphs, links, works, raw

graphs, links, works, raw = load_all(LANGS)
```

## Descriptive comparison of the two networks

How do the two transmission networks compare in shape? Density and clustering fall naturally as a network grows, so the two columns are read qualitatively rather than as a strict ratio.

```python
def network_summary(G, df, w):
    deg = dict(G.degree())
    wts = [d["weight"] for *_, d in G.edges(data=True)]
    comps = list(nx.connected_components(G))
    wpm = df.dropna(subset=["manuscript_id"]).groupby("manuscript_id")["work"].nunique()
    wits = df.groupby("work_id")["witness_id"].nunique()
    ms_date = df.dropna(subset=["manuscript_id"]).groupby("manuscript_id")["date_mid"].mean().dropna()
    work_date = pd.to_numeric(w.get("date_mid"), errors="coerce").dropna()
    return {
        "Works (nodes)": G.number_of_nodes(),
        "Co-occurring works": sum(d > 0 for d in deg.values()),
        "Manuscripts": int(df["manuscript_id"].nunique()),
        "Multi-work manuscripts": int((wpm > 1).sum()),
        "Witnesses": len(df),
        "Edges": G.number_of_edges(),
        "Density": round(nx.density(G), 4),
        "Avg degree": round(np.mean(list(deg.values())), 2),
        "Avg edge weight": round(np.mean(wts), 2) if wts else 0,
        "Max edge weight": max(wts) if wts else 0,
        "Components": len(comps),
        "Largest component": max((len(c) for c in comps), default=0),
        "Avg clustering": round(nx.average_clustering(G), 4),
        "Communities": len({G.nodes[n]["community"] for n in G.nodes()}),
        "Mean works / manuscript": round(wpm.mean(), 2),
        "Max works / manuscript": int(wpm.max()) if len(wpm) else 0,
        "Mean witnesses / work": round(wits.mean(), 2),
        "Median work date": round(work_date.median()) if len(work_date) else np.nan,
        "Median manuscript date": round(ms_date.median()) if len(ms_date) else np.nan,
    }

summary = pd.DataFrame({l: network_summary(graphs[l], links[l], works[l]) for l in LANGS})
summary.to_csv("data/network_comparison.csv")
summary
```

**Table 1.** Structural comparison of the French and German co-appearance networks: counts of works, manuscripts and links, with density, degree, clustering, fragmentation and anthology size.

## Distributions behind the summary

Works per manuscript, witnesses per work and edge weight are right-skewed and shown on box-and-jitter panels. Manuscript preservation state (fragmentary, complete, defective, …) is categorical and shown as a grouped-bar panel instead.

```python
from scipy.stats import mannwhitneyu

def num(df, c): return pd.to_numeric(df.get(c), errors="coerce").dropna()
def works_per_ms(df): return df.dropna(subset=["manuscript_id"]).groupby("manuscript_id")["work"].nunique()
def ms_dates(df): return df.dropna(subset=["manuscript_id"]).groupby("manuscript_id")["date_mid"].mean().dropna()
def wits_per_work(df): return df.groupby("work_id")["witness_id"].nunique()
def edge_weights(G): return pd.Series([d["weight"] for *_, d in G.edges(data=True)])
def work_span(df):
    s, e = num(df, "date_start"), num(df, "date_end")
    j = pd.concat([s, e], axis=1).dropna()
    return j.iloc[:, 1] - j.iloc[:, 0]

def sig_stars(p, alpha=0.05):
    if p < alpha / 1000:      # keep roughly the same visual banding, scaled to corrected alpha
        return "***"
    elif p < alpha / 10:
        return "**"
    elif p < alpha:
        return "*"
    else:
        return "ns"

panels = [
    ("Works per manuscript", lambda l: works_per_ms(links[l])),
    ("Work date (midpoint)", lambda l: num(works[l], "date_mid")),
    ("Manuscript date (midpoint)", lambda l: ms_dates(links[l])),
    ("Witnesses per work", lambda l: wits_per_work(links[l])),
    ("Work dating span (years)", lambda l: work_span(works[l])),
    ("Edge weight (shared manuscripts)", lambda l: edge_weights(graphs[l])),
]

n_tests = len(panels)
alpha_corrected = 0.05 / n_tests  # Bonferroni-corrected significance threshold
print(f"Bonferroni-corrected alpha ({n_tests} tests): {alpha_corrected:.5f}")

rng = np.random.default_rng(0)
fig, axes = plt.subplots(2, 3, figsize=(13, 7.5))
results = []
for ax, (title, getter) in zip(axes.ravel(), panels):
    data = [pd.Series(getter(l)).dropna().astype(float).values for l in LANGS]

    bp = ax.boxplot(data, patch_artist=True, widths=0.5, showfliers=False,
                    medianprops=dict(color=INK, lw=1.4),
                    whiskerprops=dict(color=SUBTLE, lw=0.8),
                    capprops=dict(color=SUBTLE, lw=0.8),
                    boxprops=dict(color=SUBTLE, lw=0.8))
    for patch in bp["boxes"]:
        patch.set_facecolor("white")
    for i, d in enumerate(data, 1):
        ax.scatter(rng.normal(i, 0.05, len(d)), d, s=6, color=INK, alpha=0.25, edgecolors="none", zorder=3)

    # --- Mann-Whitney U test + Bonferroni-corrected significance bracket ---
    statistic, p_value = mannwhitneyu(data[0], data[1], alternative="two-sided")
    sig_text = sig_stars(p_value, alpha=alpha_corrected)
    results.append(dict(panel=title, statistic=statistic, p_value=p_value,
                        p_corrected_threshold=alpha_corrected, sig=sig_text))

    y_max = max(d.max() for d in data)
    y_min = min(d.min() for d in data)
    y_range = y_max - y_min if y_max > y_min else y_max
    bracket_height = y_max + y_range * 0.08
    line_height = bracket_height + y_range * 0.03

    ax.plot([1, 1], [bracket_height, line_height], color=INK, linewidth=0.8)
    ax.plot([2, 2], [bracket_height, line_height], color=INK, linewidth=0.8)
    ax.plot([1, 2], [line_height, line_height], color=INK, linewidth=0.8)
    ax.text(1.5, line_height + y_range * 0.02, sig_text,
            ha="center", va="bottom", fontsize=10, color=INK)
    ax.set_ylim(top=line_height + y_range * 0.15)

    style_ax(ax)
    ax.set_xticks([1, 2]); ax.set_xticklabels([l.capitalize() for l in LANGS], fontsize=9)
    ax.set_title(title, fontsize=10, color=SUBTLE, pad=8)

fig.suptitle("French vs German - descriptive comparison (Bonferroni-corrected)", fontsize=12, color=SUBTLE, y=1.0)
plt.tight_layout()
plt.show()

results_df = pd.DataFrame(results)
results_df
```

## Matiere distribution by language

Share of works assigned to each matiere, all categories including Other/Unknown, for context before the matiere-restricted analyses below. 
For German, the part of *Other* that is *Heldendichtung* is shown as a purple sub-bar at the tip of the *Other* bar, with its absolute count; 
this relies on the `is_heldenepik`/`mat5` tags added at load time (cell *Loading both traditions*), so it is populated before the German-specific split analysis further down.

```python
from matplotlib.patches import Patch

HELDEN_COLOR = "#6A4C93"   # matches the Heldendichtung colour used in the network figures

def _is_german(lang):
    return lang.lower().startswith("german") or lang.lower() == "de"

def matiere_distribution(G, order=MAT_ORDER_ALL):
    counts = Counter(nx.get_node_attributes(G, "matiere").values())
    total = sum(counts.values())
    pct = pd.Series({c: counts.get(c, 0) / total * 100 for c in order}, index=order)
    raw = pd.Series({c: counts.get(c, 0) for c in order}, index=order)
    return pct, raw

def heldendichtung_count(G):
    """Works that are Heldendichtung (a subset of Other). Prefers mat5, falls
    back to an is_heldenepik flag among Other-labelled works."""
    m5 = nx.get_node_attributes(G, "mat5")
    if m5:
        return sum(v == "Heldendichtung" for v in m5.values())
    flag = nx.get_node_attributes(G, "is_heldenepik")
    mat = nx.get_node_attributes(G, "matiere")
    return sum(bool(flag.get(n)) and mat.get(n) == "Other" for n in G.nodes())

def plot_matiere_comparison(graphs, langs):
    l1, l2 = langs
    pct1, raw1 = matiere_distribution(graphs[l1])
    pct2, raw2 = matiere_distribution(graphs[l2])
    cats = pct1.index.tolist(); y = np.arange(len(cats))
    totals = {l1: raw1.sum(), l2: raw2.sum()}

    fig, ax = plt.subplots(figsize=(10, 5))
    ax.barh(y, -pct1.values, color=[MAT_COLORS[c] for c in cats],
            edgecolor="white", linewidth=0.6, height=0.65, alpha=0.9)
    ax.barh(y, pct2.values, color=[MAT_COLORS[c] for c in cats],
            edgecolor="white", linewidth=0.6, height=0.65, alpha=0.9, hatch="///")

    # --- Heldendichtung sub-bar at the tip of the Other bar, German side only ---
    has_helden = False
    for lang, sign, hatch in [(l1, -1, None), (l2, +1, "///")]:
        if not _is_german(lang) or "Other" not in cats:
            continue
        n_h = heldendichtung_count(graphs[lang])
        if n_h == 0:
            continue
        has_helden = True
        yi = cats.index("Other")
        other_pct = (pct1 if sign < 0 else pct2)["Other"]
        helden_pct = n_h / totals[lang] * 100
        start, end = sign * (other_pct - helden_pct), sign * other_pct   # outer (tip) slice of Other
        ax.barh(yi, abs(end - start), left=min(start, end), color=HELDEN_COLOR,
                edgecolor="white", linewidth=0.6, height=0.65, alpha=0.95, hatch=hatch)
        ax.text((start + end) / 2, yi - 0.46, f"Heldend. {n_h}", ha="center", va="top",
                fontsize=7.5, color=HELDEN_COLOR, fontweight="bold")

    for yi, p, n in zip(y, pct1.values, raw1.values):
        ax.text(-p - 1, yi, f"{p:.0f}% ({int(n)})", ha="right", va="center", fontsize=9, color="#333333")
    for yi, p, n in zip(y, pct2.values, raw2.values):
        ax.text(p + 1, yi, f"{p:.0f}% ({int(n)})", ha="left", va="center", fontsize=9, color="#333333")

    ax.set_yticks(y); ax.set_yticklabels(cats, fontsize=10)
    ax.set_xticks([]); ax.axvline(0, color="#333333", linewidth=0.8)

    xmax = max(pct1.max(), pct2.max()) * 1.6
    ax.set_xlim(-xmax, xmax)
    for spine in ["top", "right", "left", "bottom"]:
        ax.spines[spine].set_visible(False)

    ax.text(-xmax * 0.98, len(cats) - 0.3, l1.capitalize(), fontsize=12, fontweight="bold", color="#333333", ha="left")
    ax.text(xmax * 0.98, len(cats) - 0.3, l2.capitalize(), fontsize=12, fontweight="bold", color="#333333", ha="right")

    if has_helden:
        ax.legend(handles=[Patch(facecolor=HELDEN_COLOR, edgecolor="white",
                                 label="Heldendichtung (subset of Other)")],
                  loc="lower center", bbox_to_anchor=(0.5, -0.12), frameon=False, fontsize=9)

    ax.set_title("Matiere distribution by language (% of works)", fontsize=13, color="#333333", pad=15)
    plt.tight_layout(); plt.show()

plot_matiere_comparison(graphs, LANGS)
```

## Networks coloured by matiere

From here on, analyses are restricted to the matieres Britain, France and Rome; Other and Unknown works are dropped. Node size grows with degree; edge width with shared-manuscript count. If matiere shapes co-transmission, same-matiere nodes should gather rather than scatter.

```python
MIN_DEGREE = {"french": 1, "german": 1}

def restrict(G, order=MAT_ORDER):
    return G.subgraph([n for n in G.nodes() if G.nodes[n]["matiere"] in order]).copy()

def plot_matiere_network(ax, G, min_degree, seed=42, weight="weight", curvature=0.15):
    H = restrict(G)
    if min_degree is not None:
        H = H.subgraph([n for n, d in H.degree() if d >= min_degree]).copy()
    pos = nx.spring_layout(H, k=1.0, iterations=300, weight=weight, seed=seed)
    node_size = [40 + 16 * H.degree(n) for n in H.nodes()]

    nx.draw_networkx_edges(H, pos, ax=ax, edge_color="#B8B8D1", alpha=0.35,
                           width=[0.6 + 0.5 * H[u][v][weight] for u, v in H.edges()],
                           connectionstyle=f"arc3,rad={curvature}", arrows=True, arrowstyle="-",
                           node_size=node_size, min_source_margin=0, min_target_margin=0)
    for cat in MAT_ORDER:
        ns = [n for n in H.nodes() if H.nodes[n]["matiere"] == cat]
        if not ns:
            continue
        nx.draw_networkx_nodes(H, pos, ax=ax, nodelist=ns, node_color=MAT_COLORS[cat],
                               node_size=[40 + 16 * H.degree(n) for n in ns],
                               linewidths=0.6, edgecolors="white", alpha=0.9)
    ax.set_axis_off()

fig, axes = plt.subplots(1, 2, figsize=(14, 7))
for ax, l in zip(axes, LANGS):
    plot_matiere_network(ax, graphs[l], MIN_DEGREE[l])
    ax.set_title(l.capitalize(), fontsize=12, color="#333333")
handles = [Line2D([0], [0], marker="o", color="w", markerfacecolor=MAT_COLORS[c],
                  markeredgecolor="white", markersize=9, label=c) for c in MAT_ORDER]
fig.legend(handles=handles, title="Matiere", loc="lower center", ncol=3,
           frameon=False, fontsize=9, bbox_to_anchor=(0.5, -0.02))
plt.tight_layout()
plt.show()
```

**Figure 2.** Co-appearance networks restricted to Britain (blue), France (red) and Rome (orange). Node size ∝ degree, edge width ∝ shared-manuscript count, both traditions shown at degree ≥ 1.

### Node-labelled networks

The same restricted networks, one language at a time, with every node labelled by work title for identification.

```python
def plot_labeled_network(G, min_degree, title, seed=42, weight="weight", curvature=0.15, figsize=(9, 9)):
    H = restrict(G)
    if min_degree is not None:
        H = H.subgraph([n for n, d in H.degree() if d >= min_degree]).copy()
    pos = nx.spring_layout(H, k=1.2, iterations=300, weight=weight, seed=seed)
    node_size = [40 + 16 * H.degree(n) for n in H.nodes()]

    fig, ax = plt.subplots(figsize=figsize)
    nx.draw_networkx_edges(H, pos, ax=ax, edge_color="#B8B8D1", alpha=0.35,
                           width=[0.6 + 0.5 * H[u][v][weight] for u, v in H.edges()],
                           connectionstyle=f"arc3,rad={curvature}", arrows=True, arrowstyle="-",
                           node_size=node_size, min_source_margin=0, min_target_margin=0)
    nx.draw_networkx_nodes(H, pos, ax=ax, node_size=node_size,
                           node_color=[MAT_COLORS[H.nodes[n]["matiere"]] for n in H.nodes()],
                           linewidths=0.4, edgecolors="white", alpha=0.9)
    nx.draw_networkx_labels(H, pos, ax=ax, font_size=7, font_color=INK)
    ax.set_title(title, fontsize=12, color="#333333")
    ax.set_axis_off()
    plt.tight_layout()
    plt.show()

for l in LANGS:
    plot_labeled_network(graphs[l], MIN_DEGREE[l], l.capitalize())
```

**Figure 2b.** Node-labelled version of the restricted network for each tradition, enabling identification of individual works.

## Does shared matiere drive co-appearance?

Restricted to Britain/France/Rome as above. `matiere_assortativity` reports the share of links (and of link weight) staying within a matiere, the same share expected by chance from 1,000 label permutations with a permutation p-value, Newman's assortativity coefficient, and agreement with the data-driven communities (ARI/AMI). The permutation test is the assumption-light check: it holds the network fixed and asks only whether the observed sorting could arise by chance.

```python
def matiere_assortativity(G, weight="weight", n_perm=1000, seed=0, order=MAT_ORDER):
    rng = np.random.default_rng(seed)
    H = restrict(G, order)
    lab = {n: H.nodes[n]["matiere"] for n in H.nodes()}
    E = list(H.edges(data=True))
    if not E or len(set(lab.values())) < 2:
        return {"Nodes": H.number_of_nodes(), "Edges": len(E)}

    def concord(lm):
        cc = cw = tot = 0
        for u, v, d in E:
            w = d.get(weight, 1); tot += w
            if lm[u] == lm[v]:
                cw += w; cc += 1
        return cc / len(E), cw / tot

    obs_u, obs_w = concord(lab)
    keys, vals = list(lab), list(lab.values())
    null = np.array([concord(dict(zip(keys, rng.permutation(vals))))[1] for _ in range(n_perm)])
    p = (1 + int(np.sum(null >= obs_w))) / (n_perm + 1)
    mat = [lab[n] for n in H.nodes()]
    com = [G.nodes[n].get("community", -1) for n in H.nodes()]
    return {
        "Nodes": H.number_of_nodes(),
        "Edges": len(E),
        "Same-matiere links": round(obs_u, 3),
        "Same-matiere links (weighted)": round(obs_w, 3),
        "Expected if random": round(float(null.mean()), 3),
        "Permutation p": round(float(p), 4),
        "Assortativity": round(float(nx.attribute_assortativity_coefficient(H, "matiere")), 3),
        "ARI vs communities": round(adjusted_rand_score(mat, com), 3),
        "AMI vs communities": round(adjusted_mutual_info_score(mat, com), 3),
    }

assortativity = pd.DataFrame({l: matiere_assortativity(graphs[l]) for l in LANGS})
assortativity
```

**Table 2.** Matiere-sorting of co-appearance links (Britain/France/Rome only): observed within-matiere share (unweighted and weighted), chance expectation from 1,000 label permutations, permutation p-value, Newman assortativity, and agreement with the data-driven communities.

## Which matieres drive it, and how the traditions differ

Assortativity is one number for the whole network; it hides which matieres cohere and where the traditions part ways. For each work we compute the share of its (weighted) links staying within its own matiere, then summarise by tradition and matiere.

```python
def matiere_node_table(G, language, weight="weight", order=MAT_ORDER):
    known = {n for n in G.nodes() if G.nodes[n]["matiere"] in order}
    rows = []
    for n in known:
        within = total = 0
        for nb in G.neighbors(n):
            if nb not in known:
                continue
            w = G[n][nb].get(weight, 1); total += w
            if G.nodes[nb]["matiere"] == G.nodes[n]["matiere"]:
                within += w
        if total > 0:
            rows.append({"work": n, "matiere": G.nodes[n]["matiere"],
                         "language": language, "within": int(within), "total": int(total)})
    return pd.DataFrame(rows)

node_df = pd.concat([matiere_node_table(graphs[l], l) for l in LANGS], ignore_index=True)

cohesion = node_df.groupby(["language", "matiere"]).agg(
    works=("within", "size"), within=("within", "sum"), total=("total", "sum"))
cohesion["within share"] = (cohesion["within"] / cohesion["total"]).round(3)
cohesion[["works", "within share"]]
```

**Table 3.** Within-matiere cohesion by tradition (Britain/France/Rome only): number of works and share of their link weight staying inside the matiere.

### Modelling the difference

A binomial model puts credible intervals on those shares. Three are fit: a null (intercept-only) model; an additive model (`matiere + language`), which assumes the French-German gap is identical for every matiere; and an interaction model (`matiere:language`), which lets each matiere differ by tradition.

Each shared-manuscript link touches two works and is counted once for each, so the binomial trials are not fully independent and the credible intervals are somewhat narrower than they should be. The permutation test above, which counts each link once, is the honest significance check; the model here describes and visualises the matiere-by-tradition pattern rather than adding fresh significance claims.

```python
shared = sorted(set(node_df.loc[node_df.language == "french", "matiere"]) &
                set(node_df.loc[node_df.language == "german", "matiere"]))
model_df = node_df[node_df["matiere"].isin(shared)].copy()

print(f"Shared matiere categories: {shared}")
model_df.groupby(["language", "matiere"]).size()
```

```python
import bambi as bmb
import arviz as az
import pymc as pm
import pytensor
pytensor.config.cxx = ""

model_null = bmb.Model("p(within, total) ~ 1", model_df, family="binomial")
idata_null = model_null.fit(draws=2000, tune=1000, chains=4, target_accept=0.9, random_seed=0)
pm.compute_log_likelihood(idata_null, model=model_null.backend.model)

m_add = bmb.Model("p(within, total) ~ 0 + matiere + language", model_df, family="binomial")
idata_add = m_add.fit(draws=2000, tune=1000, chains=4, target_accept=0.9, random_seed=0)
pm.compute_log_likelihood(idata_add, model=m_add.backend.model)

model_int = bmb.Model("p(within, total) ~ 0 + matiere:language", model_df, family="binomial")
idata_int = model_int.fit(draws=2000, tune=1000, chains=4, target_accept=0.9, random_seed=0)
pm.compute_log_likelihood(idata_int, model=model_int.backend.model)

compare_dict = {
    "null (intercept only)": idata_null,
    "additive (matiere + language)": idata_add,
    "interaction (matiere:language)": idata_int,
}
```

```python
comparison_df = az.compare(compare_dict)#, reference="null (intercept only)")
print(comparison_df)

import arviz_plots as azp
azp.plot_compare(comparison_df)
plt.tight_layout()
plt.show()
```

```python
from scipy.special import expit

da = idata_int.posterior["matiere:language"]
dim = [d for d in da.dims if d not in ("chain", "draw")][0]
P = expit(da.stack(s=("chain", "draw")).transpose(dim, "s").values)

pred = []
for label, samp in zip((str(x) for x in da[dim].values), P):
    mat, lang = [t.strip() for t in label.split(",")]
    pred.append(dict(matiere=mat, language=lang, mean=samp.mean(),
                     lo89=np.quantile(samp, .055), hi89=np.quantile(samp, .945),
                     lo50=np.quantile(samp, .25), hi50=np.quantile(samp, .75)))
pred = pd.DataFrame(pred)

LANG_STYLE = {"french": dict(marker="o", fillstyle="full"), "german": dict(marker="s", fillstyle="none")}

fig, ax = plt.subplots(figsize=(8, 5))
for j, mat in enumerate(MAT_ORDER):
    col = MAT_COLORS[mat]
    for k, lang in enumerate(LANGS):
        r = pred[(pred.matiere == mat) & (pred.language == lang)]
        if r.empty:
            continue
        r = r.iloc[0]; x = j + (k - 0.5) * 0.34
        lstyle = LANG_STYLE[lang]
        ax.plot([x, x], [r.lo89, r.hi89], color=col, lw=1.0, zorder=2)
        ax.plot([x, x], [r.lo50, r.hi50], color=col, lw=3.2, alpha=0.45, solid_capstyle="butt", zorder=2)
        ax.plot(x, r["mean"], marker=lstyle["marker"], color=col,
                markerfacecolor=col if lstyle["fillstyle"] == "full" else "white",
                markeredgecolor=col, markeredgewidth=1.4, ms=8, zorder=3,
                linestyle="none", label=lang.capitalize() if j == 0 else "")

chance = 1 / len(MAT_ORDER)
ax.axhline(chance, color="#B0B0B0", lw=0.8, ls=(0, (4, 3)), alpha=0.7)
ax.text(len(MAT_ORDER) - 0.5, chance + 0.01, "~ random", fontsize=7.5, color="#8A8A8A", va="bottom", ha="right", alpha=0.8)
ax.set_xticks(range(len(MAT_ORDER))); ax.set_xticklabels(MAT_ORDER, fontsize=9)
ax.set_ylim(0, 1); ax.set_ylabel("P(neighbour shares matiere)", fontsize=9, color="#333333")
style_ax(ax)
ax.legend(frameon=False, fontsize=9, loc="upper right")
ax.set_title("Within-matiere cohesion by tradition (posterior)", fontsize=11, color="#333333", pad=10)
plt.tight_layout()
plt.show()
```

**Figure 3.** Posterior probability that a work's manuscript-neighbour shares its matiere, by matiere and tradition (French circle, German square), restricted to Britain/France/Rome. Thick bar = 50% credible interval, thin line = 89%. Dashed line marks the chance level (1/3, three matieres).

The null model, `p(within, total) ~ 1`, estimates a single homophily rate pooled across matiere and language, assuming no distinction between categories or traditions. It is the baseline against which the matiere- and language-structured models are compared: if those richer models do not out-predict it, there is no evidence that matiere or language shapes co-occurrence beyond a generic tendency toward homophily.

## German only — does splitting *Other* into *Other* vs *Heldendichtung* help explain the network?

Everything above treats *Other* as a single residual bucket and drops it from the structural tests. The German works file now carries an `is_heldenepik` flag marking works of *Heldendichtung* (heroic epic — the Dietrich and Nibelungen cycles, *Kudrun*, *Ortnit/Wolfdietrich*, and the rest), a distinctively German grouping that lives almost entirely inside *Other*.

This section asks a single question, **for German only**: is *Other* internally heterogeneous in a way the co-appearance network can see? Concretely, does carving *Heldendichtung* out of *Other* yield a node labelling that matches the co-transmission structure **better** than leaving *Other* whole? Because *Heldendichtung* is meant as a subdivision **of** *Other*, the label is applied only to flagged works whose matiere resolves to *Other*; Britain/France/Rome are left untouched. (One flagged work, *Alpharts Tod*, resolves to Britain via its `matter_local`; it is reported separately and kept as Britain rather than silently recoded.)

All structural measures below run on the **co-appearing** German subgraph (works with at least one shared-manuscript link), across all four matieres — *Other* is kept here, unlike Figures 2–3.

```python
from networkx.algorithms.community import modularity as nx_modularity

HELDEN = "Heldendichtung"
HELDEN_COLOR = "#6A4C93"                              # distinct purple, clear of Other's teal
GER_MAT4 = ["Britain", "France", "Rome", "Other"]                 # Other kept whole
GER_MAT5 = ["Britain", "France", "Rome", "Other", HELDEN]         # Other split
GER_COLORS = {**MAT_COLORS, HELDEN: HELDEN_COLOR}

def german_structure(G, w, id_col="work_id", flag_col="is_heldenepik", min_degree=1):
    """Co-appearing German subgraph (degree >= min_degree), tagged with two labellings:
       mat4 = matiere with Other whole; mat5 = Heldendichtung carved out of Other only.
       Returns (subgraph, list-of-stray-works) where strays are flagged but not Other."""
    # pd.notna guard: a missing flag must read as False, not as bool(nan) == True
    flag = ({int(r[id_col]): bool(r[flag_col]) and pd.notna(r[flag_col])
             for _, r in w.iterrows()} if flag_col in w.columns else {})
    H = G.subgraph([n for n in G.nodes() if G.nodes[n]["matiere"] in GER_MAT4])
    if min_degree:
        H = H.subgraph([n for n, d in H.degree() if d >= min_degree])
    H = H.copy()
    stray = []
    for n in H.nodes():
        h = flag.get(H.nodes[n].get("work_id"), False)
        m = H.nodes[n]["matiere"]
        H.nodes[n]["is_heldenepik"] = h
        H.nodes[n]["mat4"] = m
        if h and m == "Other":
            H.nodes[n]["mat5"] = HELDEN
        else:
            H.nodes[n]["mat5"] = m
            if h and m != "Other":
                stray.append(n)
    return H, stray

def _partition(H, attr, order):
    g = {}
    for n in H.nodes():
        if H.nodes[n][attr] in order:
            g.setdefault(H.nodes[n][attr], set()).add(n)
    return list(g.values())

def _within_share(H, attr, weight="weight"):
    cw = tot = 0
    for u, v, d in H.edges(data=True):
        wt = d.get(weight, 1); tot += wt
        if H.nodes[u][attr] == H.nodes[v][attr]:
            cw += wt
    return cw / tot if tot else np.nan

def partition_quality(H, attr, order, weight="weight"):
    comms = _partition(H, attr, order)
    mat = [H.nodes[n][attr] for n in H.nodes()]
    com = [H.nodes[n].get("community", -1) for n in H.nodes()]
    return {
        "Labels": len(comms),
        "Modularity Q": round(nx_modularity(H, comms, weight=weight), 3),
        "Assortativity": round(nx.attribute_assortativity_coefficient(H, attr), 3),
        "Same-label links (wt)": round(_within_share(H, attr, weight), 3),
        "ARI vs communities": round(adjusted_rand_score(mat, com), 3),
        "AMI vs communities": round(adjusted_mutual_info_score(mat, com), 3),
    }

Hde, stray = german_structure(graphs["german"], works["german"])
n_helden = sum(Hde.nodes[n]["mat5"] == HELDEN for n in Hde.nodes())
print(f"Co-appearing German works: {Hde.number_of_nodes()} nodes, {Hde.number_of_edges()} edges")
print(f"Heldendichtung nodes (carved from Other): {n_helden}")
if stray:
    print(f"Flagged as Heldenepik but not resolving to Other (kept as-is): {stray}")

quality = pd.DataFrame({
    "Other whole (4 matieres)":      partition_quality(Hde, "mat4", GER_MAT4),
    "Other split (+ Heldendichtung)": partition_quality(Hde, "mat5", GER_MAT5),
}).loc[["Labels", "Modularity Q", "Assortativity",
        "Same-label links (wt)", "ARI vs communities", "AMI vs communities"]]
quality
```

**Table 4.** Partition quality of the co-appearing German network under two matiere labellings: *Other* left whole (4 groups) vs *Heldendichtung* carved out of it (5 groups). *Modularity Q* rewards a labelling only when within-group links exceed the degree-based chance expectation, so — unlike the raw same-label share, which can only fall when a group is subdivided — it can rise under a **meaningful** split. *Assortativity* is Newman's coefficient for the labelling; *ARI/AMI* measure agreement with the data-driven greedy-modularity communities.

### Is the gain real, or an artefact of splitting?

Any split changes the numbers; the question is whether *this* split, along the *Heldendichtung* line, beats an arbitrary split of *Other* of the same size. Two assumption-light permutation tests, in the spirit of Table 2:

1. **Modularity gain (ΔQ).** Hold the network fixed and compare the observed ΔQ = Q(split) − Q(whole) against the ΔQ from 1,000 random relabellings that recolour the *same number* of *Other* works as *Heldendichtung*. If the real grouping tracks co-transmission, its ΔQ sits in the right tail.
2. **Within-*Other* cohesion.** Restrict to the *Other* subgraph and ask whether *Heldendichtung* works co-appear with each other more than a random binary label of the same frequency would — the targeted, one-link-per-edge check that *Heldendichtung* is itself a cohesive block.

```python
def split_gain_test(H, n_perm=1000, seed=0, weight="weight"):
    """Permutation test: observed modularity gain from the Heldendichtung split vs the gain
    from random size-matched relabellings of Other."""
    rng = np.random.default_rng(seed)
    q4 = nx_modularity(H, _partition(H, "mat4", GER_MAT4), weight=weight)
    q5 = nx_modularity(H, _partition(H, "mat5", GER_MAT5), weight=weight)
    obs = q5 - q4
    other = [n for n in H.nodes() if H.nodes[n]["mat4"] == "Other"]
    k = sum(1 for n in other if H.nodes[n]["mat5"] == HELDEN)
    orig = {n: H.nodes[n]["mat5"] for n in other}
    null = np.empty(n_perm)
    try:
        for i in range(n_perm):
            chosen = set(rng.choice(other, size=k, replace=False))
            for n in other:
                H.nodes[n]["mat5"] = HELDEN if n in chosen else "Other"
            null[i] = nx_modularity(H, _partition(H, "mat5", GER_MAT5), weight=weight) - q4
    finally:
        for n in other:
            H.nodes[n]["mat5"] = orig[n]
    p = (1 + int(np.sum(null >= obs))) / (n_perm + 1)
    return dict(Q_whole=round(q4, 3), Q_split=round(q5, 3), deltaQ=round(obs, 3),
                null_mean=round(float(null.mean()), 3), perm_p=round(float(p), 4)), null, obs

def helden_cohesion_within_other(H, n_perm=1000, seed=1, weight="weight"):
    """Do Heldendichtung works co-appear with one another more than chance, among Other works?"""
    rng = np.random.default_rng(seed)
    O = H.subgraph([n for n in H.nodes() if H.nodes[n]["mat4"] == "Other"]).copy()
    lab = {n: O.nodes[n]["is_heldenepik"] for n in O.nodes()}
    E = list(O.edges(data=True))
    if not E or len(set(lab.values())) < 2:
        return {"note": "too few within-Other edges or one-sided label"}
    def wshare(lm):
        cw = tot = 0
        for u, v, d in E:
            wt = d.get(weight, 1); tot += wt
            if lm[u] == lm[v]:
                cw += wt
        return cw / tot
    obs = wshare(lab)
    keys, vals = list(lab), list(lab.values())
    null = np.array([wshare(dict(zip(keys, rng.permutation(vals)))) for _ in range(n_perm)])
    p = (1 + int(np.sum(null >= obs))) / (n_perm + 1)
    return {"Other works": O.number_of_nodes(), "Within-Other edges": len(E),
            "Same-label share (wt)": round(obs, 3),
            "Expected if random": round(float(null.mean()), 3),
            "Permutation p": round(float(p), 4)}

split_res, null, obs = split_gain_test(Hde, n_perm=1000)
cohesion_res = helden_cohesion_within_other(Hde, n_perm=1000)
print("Modularity-gain test:", split_res)
print("Within-Other cohesion:", cohesion_res)

fig, ax = plt.subplots(figsize=(7.5, 4.2))
ax.hist(null, bins=30, color="#B8B8D1", edgecolor="white", linewidth=0.4)
ax.axvline(0, color=SUBTLE, lw=0.8, ls=(0, (4, 3)))
ax.axvline(obs, color=HELDEN_COLOR, lw=2.2)
ax.text(obs, ax.get_ylim()[1] * 0.95,
        f"  observed ΔQ = {obs:.3f}\n  p = {split_res['perm_p']:.3f}",
        color=HELDEN_COLOR, fontsize=9, va="top", ha="left")
ax.set_xlabel("ΔQ  =  modularity(Other split)  −  modularity(Other whole)", fontsize=9, color=SUBTLE)
ax.set_ylabel("random size-matched splits of Other", fontsize=9, color=SUBTLE)
style_ax(ax)
ax.set_title("Does carving Heldendichtung out of Other improve the partition?",
             fontsize=11, color=SUBTLE, pad=8)
plt.tight_layout()
plt.show()
```

**Figure 4.** Permutation null for the modularity gain. The histogram is ΔQ from 1,000 random relabellings that recolour the same number of *Other* works as *Heldendichtung*; the vertical line is the observed *Heldendichtung* split. A random subdivision of a cohesive bucket typically *lowers* modularity (mass left of 0), so an observed ΔQ in the right tail (small *p*) means the real grouping captures co-transmission structure that an arbitrary split does not.
